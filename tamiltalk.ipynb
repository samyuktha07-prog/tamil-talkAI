```python


import gradio as gr
import re

TAMIL_WORDS = {
    "naan": "நான்",
    "nee": "நீ",
    "neenga": "நீங்கள்",
    "enna": "என்ன",
    "ennaiku": "என்னைக்கு",
    "iniku": "இன்னைக்கு",
    "naalaiku": "நாளைக்கு",
    "vandhuten": "வந்துவிட்டேன்",
    "varen": "வருகிறேன்",
    "poren": "போகிறேன்",
    "poganum": "போகணும்",
    "poga": "போக",
    "veetuku": "வீட்டுக்கு",
    "college": "கல்லூரி",
    "school": "பள்ளி",
    "saaptiya": "சாப்பிட்டாயா",
    "saapten": "சாப்பிட்டேன்",
    "nalla": "நல்ல",
    "iruken": "இருக்கிறேன்",
    "irukiya": "இருக்கிறாயா",
    "venum": "வேண்டும்",
    "vendam": "வேண்டாம்",
    "help": "உதவி",
    "romba": "ரொம்ப",
    "nandri": "நன்றி",
    "vanakkam": "வணக்கம்",
    "amma": "அம்மா",
    "appa": "அப்பா",
    "friend": "நண்பர்",
    "thanni": "தண்ணீர்",
    "sapadu": "சாப்பாடு",
    "enakku": "எனக்கு",
    "ungalukku": "உங்களுக்கு",
    "pidikkum": "பிடிக்கும்",
    "theriyum": "தெரியும்",
    "illa": "இல்லை",
    "aama": "ஆமாம்",
    "yen": "ஏன்",
    "epdi": "எப்படி",
    "enga": "எங்கே",
    "ippo": "இப்போது",
    "aprom": "அப்புறம்",
    "konjam": "கொஞ்சம்",
    "pesu": "பேசு",
    "tamil": "தமிழ்",
    "english": "ஆங்கிலம்",
    "late": "தாமதம்",
    "happy": "மகிழ்ச்சி",
}

ENGLISH_MEANINGS = {
    "naan college ku poren": "I am going to college.",
    "naan nalla iruken": "I am fine.",
    "enakku help venum": "I need help.",
    "nee epdi iruka": "How are you?",
    "romba nandri": "Thank you very much.",
    "vanakkam": "Hello.",
    "naan saapten": "I have eaten.",
    "naalaiku varen": "I will come tomorrow.",
    "enakku tamil pidikkum": "I like Tamil.",
    "thanni venum": "I need water.",
}

def analyze(text, output_language):
    text = (text or "").strip()

    if not text:
        return "", "", "Please enter a sentence."

    original = text


    normalized = re.sub(r"\s+", " ", text.lower()).strip()
    normalized = re.sub(r"[.!?,]+$", "", normalized)


    english = ENGLISH_MEANINGS.get(
        normalized,
        "Phrase not available in the offline English dictionary."
    )

    words = re.findall(r"\w+|[^\w\s]", text.lower())
    converted = []

    for word in words:
        if word in TAMIL_WORDS:
            converted.append(TAMIL_WORDS[word])
        else:
            converted.append(word)

    tamil = " ".join(converted)
    tamil = re.sub(r"\s+([.,!?])", r"\1", tamil)

    if output_language == "Tamil":
        result = tamil
    elif output_language == "English":
        result = english
    else:
        result = (
            "This offline prototype supports Tamil conversion "
            "and a limited English phrase dictionary."
        )

    report = (
        "TAMILTALK AI ANALYSIS\n"
        "---------------------\n"
        f"Original input: {original}\n\n"
        f"Tamil conversion: {tamil}\n\n"
        f"English meaning: {english}\n\n"
        "Note: Dictionary-based results may be incomplete. "
        "This prototype does not perform speech recognition "
        "or generate audio."
    )

    return tamil, result, report

with gr.Blocks(title="TamilTalk AI") as app:
    gr.Markdown("""
    # TamilTalk AI
    ### Offline Tanglish Text Assistant

    Enter Tanglish, such as `naan college ku poren`,
    and explore Tamil word conversion.
    """)

    text = gr.Textbox(
        label="Enter Tamil, English, or Tanglish",
        placeholder="naan college ku poren",
        lines=3
    )

    language = gr.Radio(
        ["Tamil", "English"],
        value="Tamil",
        label="Output"
    )

    button = gr.Button("Analyze Text", variant="primary")

    tamil_output = gr.Textbox(label="Tamil word conversion")
    result_output = gr.Textbox(label="Selected output")
    report_output = gr.Textbox(label="Analysis report", lines=10)

    button.click(
        analyze,
        inputs=[text, language],
        outputs=[tamil_output, result_output, report_output]
    )

app.launch()
```
